# Rosette

Seventy-two squares, each turned 5° further than the one before, with the hue going once around
the color wheel. A ring of 36 circles fills the middle.

The turtle in these notebooks comes from [`nbturtle.py`](nbturtle.py), a stand-in for Python's
`turtle` module that draws into the notebook instead of opening a window. The commands are the
same: `forward`, `left`, `circle`, `pencolor`, `goto`, …

```python
from common import hsv
from nbturtle import render
```

## One square, 72 times

A square is four rounds of *forward, turn left 90°*. Before each square, `setheading` points the
turtle a little further around, so the squares fan out from the center. `hsv(i / COUNT, …)` turns
the hue once around the color wheel as `i` runs from 0 to `COUNT`.

```python
COUNT = 72
SIDE = 270


def squares(t):
    t.pensize(1.5)
    for i in range(COUNT):
        t.pencolor(hsv(i / COUNT, 0.7, 1.0))
        t.setheading(i * 360 / COUNT)
        t.pendown()
        for _ in range(4):
            t.forward(SIDE)
            t.left(90)
        t.penup()


render(squares, dpi=36)
```

## A ring of circles

`circle(95)` draws a circle whose center is 95 px to the turtle's left, so turning the turtle
before each circle swings it around the starting point. 36 circles, 10° apart, make a torus.

```python
def circles(t):
    t.pensize(1)
    for i in range(36):
        t.pencolor(hsv(0.55 + i / 72, 0.5, 0.95))
        t.setheading(i * 10)
        t.pendown()
        t.circle(95)
        t.penup()


render(circles, dpi=36)
```

## Together

```python
def draw(t):
    squares(t)
    circles(t)


render(draw)
```

## Try

- `COUNT = 12` shows the separate squares; `COUNT = 180` packs them into a solid ring.
- Swap the square for a pentagon: five sides with `t.left(72)`.
- `hsv(3 * i / COUNT, …)` goes around the color wheel three times instead of once.
